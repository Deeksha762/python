```python
# Corrected and improved groupby by height examples
# 1) Count rows for each exact height value
height_counts = bios_new.groupby('height_cm').size().reset_index(name='count')
height_counts = height_counts.sort_values('count', ascending=False)

# 2) Aggregations grouped by height: count names and (optionally) mean year if column exists
agg_cols = {'name': 'count'}
if 'year' in bios_new.columns:
    agg_cols['year'] = 'mean'
height_agg = bios_new.groupby('height_cm').agg(agg_cols).reset_index()

# 3) Grouping by categories. Use existing 'height_category' if available, otherwise create bins
if 'height_category' in bios_new.columns:
    cat_counts = bios_new.groupby('height_category').size().reset_index(name='count')
else:
    bins = [0, 165, 185, 1000]  # short, average, tall
    labels = ['Short', 'Average', 'Tall']
    bios_new['height_category_bin'] = pd.cut(bios_new['height_cm'], bins=bins, labels=labels, right=False)
    cat_counts = bios_new.groupby('height_category_bin').size().reset_index(name='count')

# Optional: show results for the first few rows of each table
print('\nCounts by exact height (top values):')
print(height_counts.head(10))
print('\nAggregations by exact height:')
print(height_agg.head(10))
print('\nCounts by height category:')
print(cat_counts)
```
